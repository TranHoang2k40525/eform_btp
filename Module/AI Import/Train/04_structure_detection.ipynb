```python
import sys
from pathlib import Path
sys.path.insert(0, '../Main')
from ai_import.config import Settings
from ai_import.excel import WorkbookAnalyzer
analyzer = WorkbookAnalyzer(Settings())
summaries = []
for path in Path('../Data/raw').glob('*.xlsx'):
    result = analyzer.analyze(str(path), preview_rows=5)
    summaries.append({'file': path.name, 'sheets': len(result.sheets), 'regions': sum(len(s.regions) for s in result.sheets), 'ms': result.elapsed_ms})
summaries
```
