# 01 — Phân tích dữ liệu (Tuần 2)
Mục tiêu: kiểm kê workbook theo metadata, không đưa dữ liệu nhạy cảm vào output. Đặt file được phép dùng trong `Data/raw/`.

```python
from pathlib import Path
from openpyxl import load_workbook
raw = Path('../Data/raw')
inventory = []
for path in sorted(raw.glob('*.xlsx')):
    wb = load_workbook(path, read_only=True, data_only=False, keep_links=False)
    inventory.append({'file': path.name, 'sheets': len(wb.sheetnames), 'names': wb.sheetnames})
    wb.close()
inventory
```

Không xuất cell value, tài khoản, token hoặc dữ liệu cá nhân. Đối chiếu kết quả với `Docs/EXCEL_ANALYSIS.md`.
